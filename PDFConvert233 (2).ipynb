```python
def extract_text_from_pdf(pdf_path):
    doc = fitz.open(pdf_path)
    text = [page.get_text("text") for page in doc]
    doc.close()
    return text
```

```python
def get_file_date(pdf_path):
    ts = os.path.getmtime(pdf_path)
    return datetime.datetime.fromtimestamp(ts)
```

```python
def pars_pdf(pages, file_dt):
    if not pages:
        return "Unknown", "", "", file_dt.strftime('%d.%m.%Y')
    full_text = " ".join(pages[:2]).replace('\xa0', ' ')
    full_text = re.sub(r'\s+', ' ', full_text)
    
    name, rating, activity, date_pub = "Unknown", "", "", ""
    
    r_match = re.search(r'(\d+[.,]\d+)\s+Общий рейтинг', full_text)
    if r_match:
        rating = r_match.group(1).replace(',', '.')

    meta_match = re.search(r'Общий рейтинг\s+(.*?)\s+(?=Подписаться|Читать|Обзор|Все потоки)', full_text)
    
    if meta_match:
        content = meta_match.group(1).strip()
        parts = content.split(' ', 1) 
        if len(parts) > 0:
            known_brands = ['Timeweb Cloud', 'Positive Technologies', 'Online patent', 'Альфа-Банк', 'ЮMoney', 'RUVDS.com']
            found_brand = False
            for brand in known_brands:
                if content.startswith(brand):
                    name = brand
                    activity = content.replace(brand, '').strip()
                    found_brand = True
                    break
            
            if not found_brand:
                name = parts[0]
                activity = parts[1] if len(parts) > 1 else ""

    if "VDS/VPS" in name or "скидка" in name.lower():
        name_fix = re.search(r'([A-Za-z0-9.-]+\.(com|ru|net))', full_text)
        if name_fix:
            name = name_fix.group(1)

    months = {'янв':1,'фев':2,'мар':3,'апр':4,'май':5,'июн':6,'июл':7,'авг':8,'сен':9,'окт':10,'ноя':11,'дек':12,
              'января':1,'февраля':2,'марта':3,'апреля':4,'мая':5,'июня':6,'июля':7,'августа':8,'сентября':9,'октября':10,'ноября':11,'декабря':12}
    
    date_match = re.search(r'(\d+)\s+(час|мин|ден|дня|дней)\w*\s+назад|вчера', full_text.lower())
    if date_match:
        found = date_match.group(0)
        n = re.findall(r'\d+', found)
        num = int(n[0]) if n else 1
        if 'час' in found: date_pub = (file_dt - timedelta(hours=num)).strftime('%d.%m.%Y')
        elif 'вчера' in found: date_pub = (file_dt - timedelta(days=1)).strftime('%d.%m.%Y')
        else: date_pub = (file_dt - timedelta(days=num)).strftime('%d.%m.%Y')
    else:
        fix = re.search(rf'(\d{{1,2}})\s+({ "|".join(months.keys()) })', full_text.lower())
        if fix:
            d, m_str = int(fix[1]), fix[2][:3]
            y = file_dt.year
            if months[m_str] > file_dt.month: y -= 1
            date_pub = f"{d:02d}.{months[m_str]:02d}.{y}"

    if not date_pub: date_pub = file_dt.strftime('%d.%m.%Y')

    return name, rating, activity, date_pub
```

```python
import fitz
import re
import os
import datetime
from datetime import timedelta

Brak = []
NameCompany = []
Raiting = []
DataPublish = []
Activity = []
TextArticle = []
Description = [] 

for pdf_path in all_pdf:
    try:  
        content_pages = extract_text_from_pdf(pdf_path)
        f_date = get_file_date(pdf_path)
        
        N, R, A, D = pars_pdf(content_pages, f_date)
        
        full_raw_text = ' '.join(content_pages).replace('\xa0', ' ').replace('\n', ' ')
        full_raw_text = re.sub(r'\s+', ' ', full_raw_text)
        content_start = re.search(r'(Подписаться|Обзор|Все потоки|Войти)\s+', full_raw_text)
        
        if content_start:
            desc_text = full_raw_text[content_start.end():].strip()
        else:
            desc_text = full_raw_text.strip()
    
        NameCompany.append(N)
        Raiting.append(R)
        DataPublish.append(D)
        Activity.append(A)
        Description.append(desc_text) 
        TextArticle.append(full_raw_text) 
    
        print(f"OK: {pdf_path.split('\\')[-1]} -> {N}")
        
    except Exception as e:
        print(f"ОШИБКА на файле {pdf_path}: {e}")
        Brak.append(pdf_path)  
    except Exception as e:
        print(f"ОШИБКА на файле {pdf_path}: {e}")
        Brak.append(pdf_path)

print(f"\nИтог: Обработано {len(NameCompany)}, Ошибок {len(Brak)}")
```
